## 2. CUSTOMERS (60 synthetic personas for the Persona Jury)

```python
# ---------- 2. CUSTOMERS (60 synthetic personas for the Persona Jury) ----------
segments = [
 ("college_student", (18, 23), ["en", "kn-en", "hi-en"], "WhatsApp/Instagram", 250, ["price", "wifi", "hangout"]),
 ("it_professional", (24, 35), ["en", "kn-en"], "Instagram", 600, ["coffee quality", "quiet", "speed"]),
 ("remote_freelancer", (25, 40), ["en"], "Google Maps", 450, ["wifi", "plugs", "long stay"]),
 ("young_family", (30, 42), ["kn", "en", "kn-en"], "WhatsApp", 900, ["kid friendly", "veg", "weekend"]),
 ("retired_regular", (58, 72), ["kn", "en"], "WhatsApp", 200, ["filter coffee", "morning", "no loud music"]),
 ("north_indian_migrant", (22, 38), ["hi", "hi-en"], "Instagram", 400, ["chai", "snacks", "veg"]),
 ("foodie_influencer", (21, 32), ["en", "hi-en"], "Instagram", 700, ["aesthetic", "new items", "photos"]),
 ("neighbour_homemaker", (35, 55), ["kn", "hi", "kn-en"], "WhatsApp", 350, ["value", "bakery", "veg"]),
]
first = ["Aarav","Diya","Rohan","Sneha","Karthik","Priya","Arjun","Ananya","Vikram","Lakshmi","Ishaan","Meghana","Rahul","Pooja","Suresh","Kavya","Nikhil","Divya","Manoj","Shreya"]
cust = []
for i in range(60):
    seg, ages, langs, channel, spend, cares = random.choice(segments)
    cust.append({
      "id": f"C{i+1:03d}", "name": random.choice(first), "segment": seg,
      "age": random.randint(*ages), "language_pref": random.choice(langs),
      "primary_channel": channel, "avg_spend_inr": int(spend * random.uniform(0.7, 1.4)),
      "visits_per_month": random.choice([1, 2, 3, 4, 6, 8, 12]),
      "veg": random.random() < 0.55, "cares_about": cares,
      "offer_sensitivity": round(random.uniform(0.2, 0.95), 2),
      "pet_peeves": random.sample(["fake urgency", "too many emojis", "robotic translation", "unclear terms", "long messages", "english-only"], 2),
      "preferred_days": random.sample(["Mon","Tue","Wed","Thu","Fri","Sat","Sun"], 2)})
dump("customers", cust)
```

## 6. HISTORICAL CAMPAIGNS (calibrate/sanity-check the Jury)

```python
# ---------- 6. HISTORICAL CAMPAIGNS (calibrate/sanity-check the Jury) ----------
hist = []
for i in range(30):
    ch = random.choice(["instagram_post","instagram_story","whatsapp_broadcast","counter_poster_a4"])
    lang = random.choice(["en","kn","hi","kn-en","hi-en"])
    base = {"instagram_post": 0.03, "instagram_story": 0.04, "whatsapp_broadcast": 0.12, "counter_poster_a4": 0.05}[ch]
    boost = (1.35 if lang in ("kn-en", "hi-en") else 1.2 if lang == "kn" else 1.0) * random.uniform(0.7, 1.4)
    hist.append({"id": f"H{i+1:02d}", "channel": ch, "language": lang, "offer_type": random.choice(["combo","bogo","pct_off","free_item"]),
                 "has_clear_price": random.random() < 0.8, "emoji_count": random.randint(0, 6), "length_words": random.randint(8, 60),
                 "reach": random.randint(300, 4000), "redemption_rate": round(base * boost, 4)})
dump("historical_campaigns", hist)
```
